# Predict Metallic Glass Properties

```python
import pandas as pd
from foundry import Foundry
from garden_ai import GardenClient

# Set DOIs for the dataset and model to use. 
# You can find these in the Web UIs
metallic_glass_dataset = "10.18126/7yg1-osf2"
metallic_glass_model = "10.26311/ep98-br79"

#Instantiate clients for Foundry-ML and Garden
f = Foundry()
g = GardenClient()

# Fetch the data from MDF/Foundry-ML
ds = f.get_dataset(doi=metallic_glass_dataset)
X,y = ds.load()['train']
comp_list = [c for c in X.iloc[0:100].Composition]

# Run the model with Garden-AI
garden = g.get_published_garden(metallic_glass_model)
result = garden.predict_metallicglass_Rc_LLM(comp_list)
result
```

# Predict Hardness of High Entropy Alloys by Composition

```python
import pandas as pd
from foundry import Foundry
from garden_ai import GardenClient

# Set DOIs for the dataset and model to use. 
# You can find these in the Web UIs
hea_dataset = "10.18126/rska-ta67"
hea_model = "10.26311/ep98-br79"

#Instantiate clients for Foundry-ML and Garden
f = Foundry(use_globus=False)
g = GardenClient()

# Fetch the data from MDF/Foundry-ML
ds = f.get_dataset(doi=hea_dataset)
X,y = ds.load()['train']
comp_list = [c for c in X.iloc[0:100].Composition]

# Run the model with Garden-AI
garden = g.get_published_garden(hea_model)
result = garden.predict_hea_hardness(comp_list)
result
```
